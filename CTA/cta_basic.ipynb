```python
import pandas as pd
```

```python
cta = pd.read_csv("cta_student_level_final.csv")
cta = cta[['state', 'grdlvl', 'race', 'sex', 'spec_speced', 'spec_gifted', 'spec_esl', 'frl', 'y_yirt' ,'xirt', 'treatment']]

cta['race'] = cta['race'].fillna("Unknown")
cta['sex'] = cta['sex'].fillna("Unknown")
cta['spec_speced'] = cta['spec_speced'].fillna("Unknown")
cta['spec_gifted'] = cta['spec_gifted'].fillna("Unknown")
cta['spec_esl'] = cta['spec_esl'].fillna("Unknown")
cta['frl'] = cta['frl'].fillna("Unknown")
```

```python
def summarize_multiple(df, cols):
    tables = []
    for col in cols:
        counts = df[col].value_counts(dropna=False)
        percents = df[col].value_counts(normalize=True, dropna=False) * 100
        
        temp = pd.DataFrame({
            'Variable': col,
            'Category': counts.index,
            'Count': counts.values,
            'Percent (%)': percents.values.round(1)
        })
        tables.append(temp)
    
    return pd.concat(tables, ignore_index=True)

summary = summarize_multiple(cta, ['state', 'grdlvl', 'race', 'sex', 'spec_speced', 'spec_gifted', 'spec_esl', 'frl'])
print(summary)
```
