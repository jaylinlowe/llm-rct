```python
import pandas as pd
```

```python
cta = pd.read_csv("cta_student_level_final.csv")
cta = cta[['state', 'grdlvl', 'race', 'sex', 'spec_speced', 'spec_gifted', 'spec_esl', 'frl', 'y_yirt' ,'xirt', 'treatment']]

cta['race'] = cta['race'].fillna("Unknown")
cta['sex'] = cta['sex'].fillna("Unknown")
cta['spec_speced'] = cta['spec_speced'].fillna("Unknown")
cta['spec_gifted'] = cta['spec_gifted'].fillna("Unknown")
cta['spec_esl'] = cta['spec_esl'].fillna("Unknown")
cta['frl'] = cta['frl'].fillna("Unknown")
```

```python
for covariate in ['state', 'grdlvl', 'race', 'sex', 'spec_speced', 'spec_gifted', 'spec_esl', 'frl' ,'xirt']:
    print(covariate)            
    print(cta[covariate].value_counts(normalize=True))
```
