```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tweets = pd.read_csv("data/twitter_raw_data_with_abstracts.csv")
```

```python
tweets2 = pd.read_csv("data/twitter_raw_data_with_extra_info.csv")
```

```python
sns.set(style="whitegrid", context="paper")
tweets['log_full_views'] = np.log(tweets['Full Text Views (e44)'] + 1)

fig, axes = plt.subplots(1, 2, figsize=(12, 4)) 

# First histogram
sns.histplot(tweets['Full Text Views (e44)'], bins=30, color='midnightblue', edgecolor='black', ax=axes[0])
axes[0].set_xlabel('Number of Views')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Untransformed')

# Second histogram
sns.histplot(tweets['log_full_views'], bins=30, color='midnightblue', edgecolor='black', ax=axes[1])
axes[1].set_xlabel('Number of Views')
axes[1].set_ylabel('Frequency')
axes[1].set_title('Log Transformed')

plt.tight_layout() 
plt.savefig('am_histograms.pdf') 
plt.show()
```
