```python
# get quick cost estimate 


import pandas as pd
from tiktoken import encoding_for_model

enc = encoding_for_model("gpt-4o-mini")

def count_tokens(text: str) -> int:
    if not isinstance(text, str):
        text = str(text)
    tokens = enc.encode(text)
    return len(tokens)

results["token_count"] = results["question"].apply(count_tokens)

results['token_count'].sum() * .15/1000000 + len(results) * 3 * 0.15/1000000
```
