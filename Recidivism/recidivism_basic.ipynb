```python
def summarize_multiple(df, cols):
    tables = []
    for col in cols:
        counts = df[col].value_counts(dropna=False)
        percents = df[col].value_counts(normalize=True, dropna=False) * 100
        
        temp = pd.DataFrame({
            'Variable': col,
            'Category': counts.index,
            'Count': counts.values,
            'Percent (%)': percents.values.round(2)
        })
        tables.append(temp)
    
    return pd.concat(tables, ignore_index=True)

summary = summarize_multiple(judges, ['nonblack', 'marijuana', 'cocaine', 'crack', 'heroin', 'pcp', 'otherdrug', 'nondrug', 'priorarr', 'priorfelarr', 'priordrugarr', 'priorfeldrugarr', 'priorcon', 'priorfelcon', 'priordrugcon', 'priorfeldrugcon', 'pwid', 'dist'])
print(summary)
```
